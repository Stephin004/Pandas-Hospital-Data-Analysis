```python
# FINAL CHALLENGE

import pandas as pd
import matplotlib.pyplot as plt

# Load the three original datasets
patients = pd.read_csv("patients.csv")
doctors = pd.read_csv("doctors.csv")
appointments = pd.read_csv("appointments.csv")

# Create the meaningful combined dataset
hospital = (
    appointments
    .merge(patients, on="Patient_ID", how="left")
    .merge(doctors, on="Doctor_ID", how="left")
)

hospital["Appointment_Date"] = pd.to_datetime(hospital["Appointment_Date"])

# Summary checks
print("Combined dataset shape:", hospital.shape)
print("\nColumns:")
print(hospital.columns.tolist())
print("\nMissing values:")
print(hospital.isna().sum())

# Create 6 meaningful visualizations
figures = []

# 1. Patients by city — business question: where is patient activity concentrated?
city_counts = patients["City"].value_counts()
fig, ax = plt.subplots(figsize=(7, 4))
city_counts.plot(kind="bar", ax=ax)
ax.set_title("Patients by City")
ax.set_xlabel("City")
ax.set_ylabel("Number of Patients")
fig.tight_layout()
p1 = "/mnt/data/hospital_01_patients_by_city.png"
fig.savefig(p1, dpi=160)
figures.append(p1)
plt.close(fig)

# 2. Patient age distribution — business question: what does the patient population look like?
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(patients["Age"], bins=5)
ax.set_title("Patient Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Number of Patients")
fig.tight_layout()
p2 = "/mnt/data/hospital_02_patient_age_distribution.png"
fig.savefig(p2, dpi=160)
figures.append(p2)
plt.close(fig)

# 3. Doctor workload — business question: which doctors handle the most appointments?
doctor_counts = hospital["Doctor_Name"].value_counts()
fig, ax = plt.subplots(figsize=(7, 4))
doctor_counts.plot(kind="bar", ax=ax)
ax.set_title("Appointments Handled by Doctor")
ax.set_xlabel("Doctor")
ax.set_ylabel("Number of Appointments")
fig.tight_layout()
p3 = "doctor_workload.png"
fig.savefig(p3, dpi=160)
figures.append(p3)
plt.close(fig)

# 4. Completed revenue by specialization — business question: which specialization generates most revenue?
specialty_revenue = (
    hospital[hospital["Status"] == "Completed"]
    .groupby("Specialization")["Consultation_Fee"]
    .sum()
    .sort_values(ascending=False)
)
fig, ax = plt.subplots(figsize=(7, 4))
specialty_revenue.plot(kind="bar", ax=ax)
ax.set_title("Completed Revenue by Specialization")
ax.set_xlabel("Specialization")
ax.set_ylabel("Completed Revenue")
fig.tight_layout()
p4 = "specialization.png"
fig.savefig(p4, dpi=160)
figures.append(p4)
plt.close(fig)

# 5. Monthly activity — business question: how is hospital activity changing over time?
monthly = hospital.groupby(hospital["Appointment_Date"].dt.to_period("M")).size()
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(monthly.index.astype(str), monthly.values, marker="o")
ax.set_title("Monthly Appointment Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Number of Appointments")
fig.tight_layout()
p5 = "appointments.png"
fig.savefig(p5, dpi=160)
figures.append(p5)
plt.close(fig)

# 6. Appointment status — business question: what proportion is completed vs cancelled?
status = hospital["Status"].value_counts()
fig, ax = plt.subplots(figsize=(6, 5))
ax.pie(status.values, labels=status.index, autopct="%1.1f%%")
ax.set_title("Appointment Status")
fig.tight_layout()
p6 = "appointment_status.png"
fig.savefig(p6, dpi=160)
figures.append(p6)
plt.close(fig)

print("\nCreated visualizations:")
for p in figures:
    print(p)

# Key numerical findings
print("\nKey findings:")
print("Total appointments:", len(hospital))
print("Completed:", (hospital["Status"] == "Completed").sum())
print("Cancelled:", (hospital["Status"] == "Cancelled").sum())
print("Completed revenue:", hospital.loc[hospital["Status"] == "Completed", "Consultation_Fee"].sum())
print("Average consultation fee:", hospital["Consultation_Fee"].mean())
print("Highest consultation fee:", hospital["Consultation_Fee"].max())
print("Lowest consultation fee:", hospital["Consultation_Fee"].min())
print("Top city:", city_counts.idxmax())
print("Top doctor by appointments:", doctor_counts.idxmax())
print("Top specialization by completed revenue:", specialty_revenue.idxmax())
print("Peak month:", monthly.idxmax())
```
